Q14 Find the second largest salary.

```python
salary = [50000,75000,62000,90000,85000]
```

```python
largest = salary[0]
second_largest = float("-inf")

for x in salary:
    if x > largest:
        second_largest = largest
        largest = x

    elif x > second_largest and x != largest:
        second_largest = x

print(second_largest)
```

Q16 You have transaction IDs:

Write a function that:

* Removes duplicates
* Preserves the original order
* Returns the cleaned list

```python
transactions = [101,102,103,101,105,102,106,107,106]
```

```python
unique = []

for x in transactions:
    if x not in unique:
        unique.append(x)

print(unique)
```

```python
def remove_dup(t_list):
    
    unique = []

    for x in t_list:
        if x not in unique:
            unique.append(x)

    return unique
```

```python
remove_dup(transactions)
```
